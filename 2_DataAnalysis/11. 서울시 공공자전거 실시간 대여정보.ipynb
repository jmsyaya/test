# **1. 서울 열린데이터 광장**
서울 열린데이터 광장(Seoul Open Data Plaza)은 서울시에서 운영하는 공공데이터 개방 플랫폼입니다. 시민, 연구자, 기업 등이 서울시에서 생성한 다양한 공공데이터를 자유롭게 활용할 수 있도록 제공하고 있습니다. 이를 통해 데이터 기반의 창의적인 아이디어와 혁신을 촉진하며, 시민들의 정보 접근성을 높이고 공공서비스를 개선하는 데 기여하고 있습니다.

### API
API(Application Programming Interface)는 서로 다른 프로그램이나 서비스가 데이터를 주고받을 수 있도록 연결해주는 통로 역할을 하는 기술입니다. 공공데이터에서 제공하는 API는 기상 정보, 교통 정보, 관광 정보, 미세먼지 정보 등 다양한 데이터를 개발자가 프로그램에서 직접 사용할 수 있도록 제공하는 서비스입니다. 예를 들어 날씨 API를 사용하면 웹사이트나 앱에서 실시간 날씨 정보를 자동으로 가져와 화면에 표시할 수 있습니다. 대부분의 공공데이터 API는 요청(Request)을 보내면 JSON 또는 XML 형태로 데이터를 응답(Response)해주며, 이를 활용하여 다양한 웹사이트, 앱, AI 서비스 등을 개발할 수 있습니다.

### JSON
JSON(JavaScript Object Notation)은 데이터를 저장하거나 프로그램끼리 데이터를 주고받기 위해 사용하는 데이터 형식입니다. 웹 개발, AI, 공공데이터 API, 앱 개발 등 거의 모든 분야에서 사용되고 있습니다.

```
{
    "name": "김사과",
    "age": 20,
    "job": "학생"
}
```

1. Python의 JSON 모듈

```
import json
```

2. Python 객체 -> JSON 문자열 변환

```
student = {"name":"김사과", "age":20, "major":"컴퓨터 공학"}
j = json.dumps(student)
```

3. JSON 문자열 -> Python 객체 변환

```
json.loads()
```

4. JSON 데이터 접근
Python에서는 딕셔너리처럼 사용
```
    print(data['name'])
```

````python
def fetch_bike_data():
    base_url = 'http://openapi.seoul.go.kr:8088/4a4e64575972797537364363425073/json/bikeList/'
    start = 1
    end = 1000
    step = 1000
    data_frames = []

    while True:
        # http://openapi.seoul.go.kr:8088/4a4e64575972797537364363425073/json/bikeList/1/1000/
        url = f"{base_url}{start}/{end}/"
        # print(url)
        response = requests.get(url)
        print(response.status_code)
        if response.status_code != 200:
            print(f"Status Code: {response.status_code}")
            break

        json_data = response.json()
        try:
            rent_bike_status = json_data["rentBikeStatus"]
            result_code = rent_bike_status["RESULT"]["CODE"]
        except KeyError:
            print("JSON 오류")
            break

        if result_code == "INFO-200":
            print("데이터 없음")
            break

````
